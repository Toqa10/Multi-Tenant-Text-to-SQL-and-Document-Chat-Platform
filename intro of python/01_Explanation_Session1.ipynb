## 2. Functions and Modular Code (~50 min)

Functions let us break code into reusable, testable pieces. Writing modular code is the
foundation of clean AI pipelines: data loading, cleaning, training, and evaluation should
each live in their own function.

### 2.1 Defining and Calling Functions

```
def greet(name, greeting="Hello"):
    """Return a greeting message."""
    return f"{greeting}, {name}!"

print(greet("Sara"))
print(greet("Omar", greeting="Hi"))
```

### 2.2 Positional vs Keyword Arguments

```
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}.")

describe_pet("dog", "Rex")          # positional
describe_pet(animal="cat", name="Milo")  # keyword
describe_pet(name="Coco", animal="parrot")  # order doesn't matter with keywords
```

### 2.3 *args and **kwargs

```
def summarize(*numbers, **details):
    print("Sum:", sum(numbers))
    for key, value in details.items():
        print(f"{key}: {value}")

summarize(1, 2, 3, author="Ahmed", topic="Math")
```

### 2.4 Return Values (single, multiple, none)

```
def min_max(numbers):
    return min(numbers), max(numbers)   # returns a tuple

low, high = min_max([4, 8, 1, 9, 3])
print("Min:", low, "Max:", high)

def log_message(msg):
    print("LOG:", msg)   # no return -> returns None

result = log_message("Training started")
print("Return value:", result)
```

### 2.5 Lambda Functions

```
square = lambda x: x ** 2
print(square(5))

# Functions as arguments (important for AI callbacks)
def apply_operation(numbers, operation):
    return [operation(n) for n in numbers]

print(apply_operation([1, 2, 3], square))
print(apply_operation([1, 2, 3], lambda x: x + 100))
```

### 2.6 Variable Scope

```
x = 10  # global variable

def modify_local():
    x = 5   # local variable, does not affect the global one
    print("Inside function:", x)

modify_local()
print("Outside function:", x)
```

### 2.7 Building a Modular Pipeline

**Modular code tip:** Split logic into small functions, each doing ONE thing.
This makes AI pipelines easier to debug, test, and reuse.

```
def load_data():
    return [1, 2, 3, 4, 5, 6, 7, 8]

def clean_data(data):
    return [x for x in data if x > 1]

def normalize_data(data):
    max_val = max(data)
    return [x / max_val for x in data]

def process_pipeline():
    data = load_data()
    cleaned = clean_data(data)
    normalized = normalize_data(cleaned)
    return normalized

print(process_pipeline())
```

## 3. List, Dictionary, and Set Comprehensions (~40 min)

Comprehensions provide a concise way to build lists, dicts, and sets in a single readable
line — very common in data preprocessing for AI.

### 3.1 List Comprehensions

```
numbers = range(10)
squares = [n**2 for n in numbers]
print(squares)

# With condition
even_squares = [n**2 for n in numbers if n % 2 == 0]
print(even_squares)

# With if/else inside the expression
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(labels)
```

### 3.2 Dictionary Comprehensions

```
words = ["ai", "python", "data"]
word_lengths = {w: len(w) for w in words}
print(word_lengths)

# Swapping keys and values
original = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in original.items()}
print(swapped)
```

### 3.3 Set Comprehensions

```
nums = [1, 2, 2, 3, 4, 4, 5]
unique_squares = {n**2 for n in nums}
print(unique_squares)
```

### 3.4 Nested Comprehensions

```
# Flattening a matrix
matrix = [[1, 2, 3], [4, 5, 6]]
flattened = [item for row in matrix for item in row]
print(flattened)

# Building a matrix
grid = [[r * c for c in range(3)] for r in range(3)]
print(grid)
```

### 3.5 Comprehensions vs Loops (performance & readability)

```
import time

n = 1_000_000

start = time.time()
result_loop = []
for i in range(n):
    result_loop.append(i * 2)
loop_time = time.time() - start

start = time.time()
result_comp = [i * 2 for i in range(n)]
comp_time = time.time() - start

print(f"Loop time: {loop_time:.4f}s")
print(f"Comprehension time: {comp_time:.4f}s")
```

## 4. Iterators and Generators (~40 min)

Generators produce values one at a time instead of storing everything in memory — critical
when working with large AI datasets.

### 4.1 Iterables vs Iterators

```
my_list = [1, 2, 3]      # an iterable
my_iter = iter(my_list)  # an iterator created from it

print(next(my_iter))
print(next(my_iter))
print(next(my_iter))
# next(my_iter)  # would raise StopIteration - nothing left
```

### 4.2 Writing a Generator Function

```
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for num in count_up_to(5):
    print(num)

gen = count_up_to(3)
print(next(gen))
print(next(gen))
```

### 4.3 Generator Expressions

```
squares_gen = (n**2 for n in range(1_000_000))  # lazy, no memory issue
print(next(squares_gen))
print(next(squares_gen))
print(type(squares_gen))
```

### 4.4 Memory Comparison: List vs Generator

```
import sys

list_version = [x for x in range(100000)]
gen_version = (x for x in range(100000))

print("List size in memory:", sys.getsizeof(list_version), "bytes")
print("Generator size in memory:", sys.getsizeof(gen_version), "bytes")
```

### 4.5 Why This Matters for AI

When reading huge datasets (millions of rows/images), generators let you process data in
batches without loading everything into RAM at once.

```
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

data = list(range(10))
for batch in batch_generator(data, 3):
    print("Batch:", batch)
```

### 4.6 Chaining Generators (a mini preprocessing pipeline)

```
def read_stream(n):
    for i in range(n):
        yield i

def filter_even(stream):
    for value in stream:
        if value % 2 == 0:
            yield value

def scale(stream, factor):
    for value in stream:
        yield value * factor

pipeline = scale(filter_even(read_stream(20)), factor=10)
print(list(pipeline))
```

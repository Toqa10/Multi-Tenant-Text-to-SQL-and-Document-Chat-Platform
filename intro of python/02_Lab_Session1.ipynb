## 2. Functions and Modular Code — Exercises (40 min)

**Task 2.1:** Write a function `celsius_to_fahrenheit(c)` that converts Celsius to Fahrenheit.

**Task 2.2:** Write a function `describe_student(name, *subjects, **extra_info)` that prints the student's name, list of subjects, and any extra info (e.g., grade, school).

**Task 2.3 (Modular pipeline):** Build a small pipeline made of 3 separate functions:
- `load_numbers()` returns a list of numbers `[4, 8, 15, 16, 23, 42]`
- `filter_even(numbers)` returns only even numbers
- `square_all(numbers)` returns each number squared

Then write a `run_pipeline()` function that calls them in order and returns the final result.


```python
# Task 2.1
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

print(celsius_to_fahrenheit(25))
```

```python
# Task 2.2
def describe_student(name, *subjects, **extra_info):
    print("Name:", name)
    print("Subjects:", subjects)
    print("Extra Info:", extra_info)

describe_student( "Toqa","OOP","Python",age=21,level=3)
```

```python
# Task 2.3
def load_numbers():
    return [4, 8, 15, 16, 23, 42]
def filter_even(numbers):
    even_numbers = []
    for num in numbers:
        if num % 2 == 0:
            even_numbers.append(num)
    return even_numbers
def square_all(numbers):
    squared_numbers = []
    for num in numbers:
        squared_numbers.append(num ** 2)
    return squared_numbers
def run_pipeline():
    numbers = load_numbers()
    even_numbers = filter_even(numbers)
    squared_numbers = square_all(even_numbers)
    return squared_numbers
print(run_pipeline())
```

## 3. Comprehensions — Exercises (40 min)

**Task 3.1:** Given `nums = list(range(1, 21))`, use a list comprehension to get all numbers divisible by 3.

**Task 3.2:** Given `words = ["data", "ai", "model", "python"]`, use a dict comprehension to map each word to `True` if its length > 3, else `False`.

**Task 3.3:** Given `nums = [1, -2, 3, -4, 5, -2, 3]`, use a set comprehension to get the set of absolute values.

**Task 3.4 (Challenge):** Given a matrix (list of lists) of numbers, use a nested list comprehension to get a new matrix where every element is doubled.


```python
# Task 3.1
nums = list(range(1, 21))
divisible_by_3 = [num for num in nums
                  if num % 3 == 0]

print(divisible_by_3)
```

```python
# Task 3.2
words = ["data", "ai", "model", "python"]
result = {
    word: len(word) > 3
    for word in words
}

print(result)
```

```python
# Task 3.3
nums = [1, -2, 3, -4, 5, -2, 3]
absolute_values = {abs(num) for num in nums}

print(absolute_values)
```

```python
# Task 3.4
matrix = [[1, 2], [3, 4], [5, 6]]
doubled_matrix = [[num * 2 for num in row] for row in matrix]

print(doubled_matrix)
```

## 4. Iterators and Generators — Exercises (40 min)

**Task 4.1:** Write a generator function `even_numbers(n)` that yields even numbers from 0 up to `n`.

**Task 4.2:** Write a generator function `fibonacci(n)` that yields the first `n` Fibonacci numbers.

**Task 4.3 (AI-related):** Write a generator function `batch_loader(data, batch_size)` that yields successive batches from a list `data`. Test it on `list(range(23))` with `batch_size=5`.


```python
# Task 4.1
def even_numbers(n):
    for num in range(0, n + 1):
        if num % 2 == 0:
            yield num
for num in even_numbers(20):
    print(num)
```

```python
# Task 4.2
def fibonacci(n):
    a = 0
    b = 1

    while n > 0:
        yield a
        c = a + b
        a = b
        b = c
        n -= 1

for num in fibonacci(10):
    print(num)
```

```python
# Task 4.3
def batch_loader(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]
data = list(range(23))
for batch in batch_loader(data, 5):
    print(batch)
```
